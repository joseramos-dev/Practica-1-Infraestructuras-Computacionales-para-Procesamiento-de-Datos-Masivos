Ahora vamos a subir los archivos .csv a HDFS, pero antes, lo trataremos para poder pasarlo a formato Parquet. Los motivos de la elección de este formato en lugar de CSV son:

- **Rendimiento:** Parquet guarda los datos por columnas, así que Hive solo lee las columnas que usa cada consulta (por ejemplo `2018`, o de `1990` a `1999`) y no las 70 columnas de cada fila. Además, los datos van comprimidos y ocupan menos espacio en HDFS.
- **Facilidad de uso:** el esquema y los tipos se guardan dentro del fichero (los años como `DOUBLE` y los valores vacíos como `NULL`). Así se evitan los problemas de leer el CSV en Hive, como los nombres de país con comas (`"Congo, Dem. Rep."`), que con un formato delimitado partirían las filas.

Antes de convertirlos, renombramos las columnas para quitar los espacios (`Country Name` -> `country_name`) y poder usarlas en Hive sin comillas.

````python
import pyarrow as pa
import pyarrow.parquet as pq

columnas = {'Country Name': 'country_name', 'Country Code': 'country_code',
            'Indicator Name': 'indicator_name', 'Indicator Code': 'indicator_code'}

df_electricidad = df_electricidad.rename(columns=columnas)
df_nuclear = df_nuclear.rename(columns=columnas)

anios = [str(a) for a in range(1960, 2026)]

esquema = pa.schema([('country_name', pa.string()), ('country_code', pa.string()),
                     ('indicator_name', pa.string()), ('indicator_code', pa.string())]
                    + [(a, pa.float64()) for a in anios])

tabla_electricidad = pa.Table.from_pandas(df_electricidad[esquema.names], schema=esquema, preserve_index=False)
pq.write_table(tabla_electricidad, 'electricidad.parq')

tabla_nuclear = pa.Table.from_pandas(df_nuclear[esquema.names], schema=esquema, preserve_index=False)
pq.write_table(tabla_nuclear, 'nuclear.parq')
````

## EJERCICIO 2.3 ##
Ahora debemos crear una vista que muestre por cada país: 
1. nombre
2. código de país
3. valor medio de la energia entre 1990-1999

````python
%%writefile crear-vista-nuclear.hql
DROP VIEW IF EXISTS vista_nuclear_90s;
CREATE VIEW vista_nuclear_90s AS
SELECT country_name, country_code,
        ( COALESCE(`1990`, 0) + COALESCE(`1991`, 0) + COALESCE(`1992`, 0) + COALESCE(`1993`, 0)
        + COALESCE(`1994`, 0) + COALESCE(`1995`, 0) + COALESCE(`1996`, 0) + COALESCE(`1997`, 0)
        + COALESCE(`1998`, 0) + COALESCE(`1999`, 0) )
        /
        NULLIF( IF(`1990` IS NOT NULL, 1, 0) + IF(`1991` IS NOT NULL, 1, 0)
              + IF(`1992` IS NOT NULL, 1, 0) + IF(`1993` IS NOT NULL, 1, 0)
              + IF(`1994` IS NOT NULL, 1, 0) + IF(`1995` IS NOT NULL, 1, 0)
              + IF(`1996` IS NOT NULL, 1, 0) + IF(`1997` IS NOT NULL, 1, 0)
              + IF(`1998` IS NOT NULL, 1, 0) + IF(`1999` IS NOT NULL, 1, 0)
        , 0) AS media_nuclear_90s
FROM nuclear;
````

````python
!beeline -u "jdbc:hive2://" -f crear-vista-nuclear.hql
!beeline -u "jdbc:hive2://" -e "show tables;"
````

````python
!beeline -u "jdbc:hive2://" -e "SELECT country_name, country_code, media_nuclear_90s FROM vista_nuclear_90s WHERE country_code IN ('ESP', 'FRA', 'USA', 'ARG');"
````

La vista calcula, para cada país, la media de la producción nuclear entre 1990 y 1999. Como cada año es una columna, no se puede usar `AVG`: se suman los años con dato (`COALESCE` convierte los `NULL` en 0) y se divide entre el número de años que no son `NULL`. `NULLIF` evita dividir entre 0 si un país no tiene ningún dato en la década.

Por ejemplo, Francia tuvo una media del 75,67 %, España del 32,95 %, Estados Unidos del 19,31 % y Argentina del 11,74 %.

2. ¿Cuál es el valor de la estadística de la electricidad en 2018, en el país que tiene el mayor valor ese año para la estadística de la energía nuclear? Primero, encuentra el país que tiene el mayor valor para la estadística de la energía nuclear en 2018 (si hay varios, devuelve el primero en orden alfabético del código del país). Luego, encuentra el valor de la estadística para electricidad en ese país en 2018. Devuelve el nombre del país, su código y el valor de la estadística para electricidad en 2018.


````python
%%writefile consulta-2-4-2.hql
SELECT e.country_name, e.country_code, e.`2018` AS electricidad_2018
FROM (
    SELECT country_code
    FROM nuclear
    WHERE `2018` IS NOT NULL
    ORDER BY `2018` DESC, country_code ASC
    LIMIT 1
) n
JOIN electricidad e ON e.country_code = n.country_code;
````

````python
!beeline -u "jdbc:hive2://" -f consulta-2-4-2.hql
````

**Respuesta:** el país con mayor porcentaje de electricidad nuclear en 2018 fue **Francia (FRA)**, y ese año el **100 %** de su población tenía acceso a la electricidad.

Una subconsulta ordena los países por el valor nuclear de 2018 (de mayor a menor, y por `country_code` en caso de empate) y se queda con el primero (`LIMIT 1`). Después se cruza con la tabla `electricidad` mediante un `JOIN` por `country_code`.

````python
%%writefile crear-tablas-olimpiadas.hql
-- tabla de atletas: fichero Parquet (formato columnar, con tipos)
DROP TABLE IF EXISTS atletas;
CREATE EXTERNAL TABLE atletas (
  id INT,
  name STRING,
  sex STRING,
  age INT,
  height DOUBLE,
  weight DOUBLE,
  team STRING,
  noc STRING,
  games STRING,
  year INT,
  season STRING,
  city STRING,
  sport STRING,
  event STRING,
  medal STRING
)
STORED AS PARQUET
LOCATION '/olimpiadas/atletas_parquet';

-- tabla de regiones: CSV original leído con OpenCSVSerde
DROP TABLE IF EXISTS regiones;
CREATE EXTERNAL TABLE regiones (
  noc STRING,
  region STRING,
  notes STRING
)
ROW FORMAT SERDE 'org.apache.hadoop.hive.serde2.OpenCSVSerde'
WITH SERDEPROPERTIES ("separatorChar" = ",", "quoteChar" = "\"")
STORED AS TEXTFILE
LOCATION '/olimpiadas/regiones'
TBLPROPERTIES ("skip.header.line.count"="1");
````

Cada tabla usa un formato de almacenamiento distinto:

- **`atletas` (Parquet):** el CSV original (40 MB, 271 116 filas) se convierte a Parquet con pandas y pyarrow. Ocupa un 86 % menos (5,7 MB), Hive solo lee las columnas que usa cada consulta y se conservan los tipos (`age` y `year` como `INT`, `height` y `weight` como `DOUBLE`, y los `NA` de `medal` como `NULL`).
- **`regiones` (CSV con `OpenCSVSerde`):** el fichero se deja como CSV y se lee con la serde `org.apache.hadoop.hive.serde2.OpenCSVSerde`. Una **serde** (serializador/deserializador) es la clase que indica a Hive cómo convertir cada línea del fichero en columnas. `OpenCSVSerde` entiende los campos entre comillas (`quoteChar`), así que una coma dentro de un campo no lo parte en dos, cosa que no hace el formato delimitado básico. Su limitación es que todas las columnas se leen como `STRING`, algo que aquí no importa porque `noc`, `region` y `notes` son texto.

Comprobamos que las tablas se han creado y cargado correctamente. Para cada tabla mostramos algunas filas y contamos el número de registros, para compararlo con los datos originales:
- `atletas` debe tener **271 116** filas, las mismas que leímos con pandas.
- `regiones` debe tener **230** filas (el CSV tiene 231 líneas y la cabecera se descarta con `skip.header.line.count`).

````python
%%writefile comprobar-tablas-olimpiadas.hql
-- Muestra de la tabla atletas (Parquet)
SELECT id, name, sex, age, height, weight, noc, year, season, sport, medal
FROM atletas
LIMIT 5;

-- Número total de filas de atletas (debe coincidir con pandas: 271116)
SELECT COUNT(*) AS filas_atletas FROM atletas;

-- Muestra de la tabla regiones (CSV leído con OpenCSVSerde)
SELECT noc, region, notes
FROM regiones
LIMIT 5;

-- Número total de filas de regiones (debe ser 230)
SELECT COUNT(*) AS filas_regiones FROM regiones;
````

````python
!beeline -u "jdbc:hive2://" -f comprobar-tablas-olimpiadas.hql
````

**Nota sobre el número de filas de `regiones`:** el `COUNT(*)` devuelve 231 filas en vez de 230. La fila de más es la cabecera del CSV (`NOC,region,notes`).

La tabla tiene definida la propiedad `skip.header.line.count = 1`, pero en este entorno (Hive 4.0 sobre MapReduce) solo se aplica en los `SELECT` simples, que Hive resuelve leyendo directamente el fichero (por eso no aparece en el `SELECT … LIMIT 5`). En las consultas que se ejecutan como un job MapReduce (`COUNT`, `GROUP BY`, `JOIN`…) la propiedad no se aplica y la cabecera se lee como una fila más.

Esto **no afecta a los resultados** de las consultas: todas cruzan `regiones` con `atletas` mediante un `JOIN` por `noc`, y ningún atleta tiene el código `'NOC'`, así que esa fila nunca participa en los resultados. Los datos se han cargado correctamente: hay 230 regiones reales más la cabecera.

### CONSULTA 1 ###
Los 5 deportes con más medallas de deportistas españoles

````python
%%writefile consulta1.hql
SELECT a.sport, COUNT(*) AS medallas
FROM atletas a
JOIN regiones r ON a.noc = r.noc
WHERE r.region = 'Spain' AND a.medal IS NOT NULL
GROUP BY a.sport
ORDER BY medallas DESC, a.sport
LIMIT 5;
````

````python
!beeline -u "jdbc:hive2://" -f consulta1.hql
````

**Resultado:** los deportes en los que deportistas españoles han ganado más medallas son el hockey (77), el baloncesto (60), el balonmano (60), el fútbol (52) y el waterpolo (36).

Todos son deportes de equipo porque el dataset tiene una fila por atleta y prueba: cuando un equipo gana una medalla, cuenta una vez por cada jugador. Por tanto, la consulta mide medallas *recibidas por deportistas*, no medallas *ganadas por el país*.

### CONSULTA 2 ###
El país con mejor ratio medallas/apariciones

````python
%%writefile consulta2.hql
SELECT r.region,
       SUM(IF(a.medal IS NOT NULL, 1, 0)) AS medallas,
       COUNT(*) AS apariciones,
       ROUND(SUM(IF(a.medal IS NOT NULL, 1, 0)) / COUNT(*), 4) AS ratio
FROM atletas a
JOIN regiones r ON a.noc = r.noc
WHERE r.region <> ''
GROUP BY r.region
HAVING COUNT(*) >= 1000
ORDER BY ratio DESC
LIMIT 5;
````

````python
!beeline -u "jdbc:hive2://" -f consulta2.hql
````

**Resultado:** Rusia es el país con mejor ratio: el 33,8 % de sus participaciones terminaron en medalla. Le siguen Estados Unidos (29,9 %), Alemania (23,7 %), Noruega (20,8 %) y Suecia (18,4 %).

Se exigen al menos 1000 participaciones (`HAVING COUNT(*) >= 1000`) para evitar que países con muy pocos deportistas salgan arriba por casualidad. Rusia incluye la Unión Soviética (URS) y el Equipo Unificado de 1992 (EUN), porque el fichero `noc_regions` asigna esos códigos a la región "Russia".

### CONSULTA 3 ###
Edad, altura y peso medios de los medallistas de oro en natación

````python
%%writefile consulta3.hql
SELECT COUNT(*) AS medallas_oro,
       ROUND(AVG(age), 1) AS edad_media,
       ROUND(AVG(height), 1) AS altura_media_cm,
       ROUND(AVG(weight), 1) AS peso_medio_kg
FROM atletas
WHERE sport = 'Swimming' AND medal = 'Gold';
````

````python
!beeline -u "jdbc:hive2://" -f consulta3.hql
````

**Resultado:** en natación se han entregado 1099 medallas de oro. Sus ganadores tenían de media 20,8 años, medían 181,7 cm y pesaban 74,0 kg. `AVG` ignora los valores `NULL`, así que los atletas sin altura o peso registrados no afectan a la media.

### CONSULTA 4 ###
Evolución del porcentaje de mujeres por año

````python
%%writefile consulta4.hql
SELECT year,
       COUNT(DISTINCT id) AS atletas,
       COUNT(DISTINCT IF(sex = 'F', id, NULL)) AS mujeres,
       ROUND(100 * COUNT(DISTINCT IF(sex = 'F', id, NULL)) / COUNT(DISTINCT id), 2) AS pct_mujeres
FROM atletas
GROUP BY year
ORDER BY year;
````

````python
!beeline -u "jdbc:hive2://" -f consulta4.hql
````

**Resultado:** el porcentaje de mujeres ha crecido de forma continua: del 0 % en 1896 al 1,9 % en 1900, el 10 % en 1948, el 25 % en 1988 y el 45 % en 2016. Se usa `COUNT(DISTINCT id)` para contar cada atleta una sola vez por año aunque participe en varias pruebas.

Desde 1994 los Juegos de verano y de invierno se celebran en años distintos, por eso los años de invierno (1994, 1998, 2002…) tienen muchos menos atletas.

### CONSULTA 5 ###
Atleta con más medallas en total

````python
%%writefile consulta5.hql
SELECT a.id, a.name, r.region,
       COUNT(*) AS medallas,
       SUM(IF(a.medal = 'Gold', 1, 0)) AS oros,
       SUM(IF(a.medal = 'Silver', 1, 0)) AS platas,
       SUM(IF(a.medal = 'Bronze', 1, 0)) AS bronces
FROM atletas a
JOIN regiones r ON a.noc = r.noc
WHERE a.medal IS NOT NULL
GROUP BY a.id, a.name, r.region
ORDER BY medallas DESC, a.name
LIMIT 5;
````

````python
!beeline -u "jdbc:hive2://" -f consulta5.hql
````

**Resultado:** el atleta con más medallas es **Michael Phelps** (Estados Unidos), con 28 (23 oros, 3 platas y 2 bronces), seguido de Larysa Latynina, con 18. Latynina aparece como "Russia" porque compitió con la Unión Soviética (URS), que el fichero de regiones asigna a Rusia.

### CONSULTA 6 ###
Países que han competido con más de un código NOC

````python
%%writefile consulta6.hql
SELECT r.region,
       COUNT(DISTINCT a.noc) AS num_codigos,
       CONCAT_WS(', ', COLLECT_SET(a.noc)) AS codigos,
       SUM(IF(a.medal IS NOT NULL, 1, 0)) AS medallas
FROM atletas a
JOIN regiones r ON a.noc = r.noc
WHERE r.region <> ''
GROUP BY r.region
HAVING COUNT(DISTINCT a.noc) > 1
ORDER BY medallas DESC
LIMIT 10;
````

````python
!beeline -u "jdbc:hive2://" -f consulta6.hql
````

**Resultado:** varios países han competido con distintos códigos a lo largo de la historia, por cambios políticos o de nombre. Por ejemplo, Alemania aparece con 4 códigos (FRG y GDR para la Alemania Occidental y Oriental, GER y SAA para el Sarre), Rusia con 3 (URS, EUN y RUS) y Serbia con 3 (YUG, SCG y SRB). `COLLECT_SET` reúne en una lista los códigos distintos de cada región y `CONCAT_WS` los une en un solo texto.
